```python
import pandas as pd
import numpy as np
```

```python
#3 sigma comparisons
def flag_3sigma(df, checks):
    """
    checks: list of (value_col, sigma_col) tuples to test
    Adds a FITFLAG column: True if any field deviates > 3 sigma
    """

    flag = pd.Series(False, index=df.index)

    for val_col, sig_col in checks:
        if val_col in df.columns and sig_col in df.columns:
            residual = np.abs(df[val_col] - df[f"{val_col}_input"]) / df[sig_col]
            flag |= (residual > 3)

    df["FITFLAG"] = flag.astype(int)
    return df
```
