```python
import pandas as pd
import numpy as np
```

```python
#3 sigma comparisons
def flag_3sigma(df, checks):
    """
    checks: list of (value_col, sigma_col) tuples to test
    Adds a FITFLAG column: True if any field deviates > 3 sigma
    """

    flag = pd.Series(False, index=df.index)

    for val_col, sig_col in checks:
        if val_col in df.columns and sig_col in df.columns:
            residual = np.abs(df[val_col] - df[f"{val_col}_input"]) / df[sig_col]
            flag |= (residual > 3)

    df["FITFLAG"] = flag.astype(int)
    return df
```

```python
data_dir = '~/BASTA/BASTA/final_runs/output/'

input_names = [
        "starid", "numax", "numax_err", "dnu", "dnu_err", "Teff", "Teff_err", "FeH", "FeH_err", "RA", "DEC", 
        "G_GAIA", "G_GAIA_err", "Mj_2MASS", "Mj_2MASS_err", "Mh_2MASS", "Mh_2MASS_err", "Mk_2MASS", "Mk_2MASS_err", "phase"
    ]
```

```python
#Table 1: numax only input
table1 = []
for letter in ['a','b','c','d']:
    input_file = pd.read_csv(f'../data/basta_ascii_files/raw_{letter}.ascii',sep='\s+',names=input_names,dtype={'starid':str})
    output_file = pd.read_csv(data_dir+f'table1_{letter}/results_clean.ascii',sep='\s+',dtype={'starid':str})

    output_df = output_file.merge(input_file,how='left',on='starid',suffixes=['','_input'])
    table1.append(output_df)
table1 = pd.concat(table1)

print(f'{table1["Teff"].isna().sum()} stars where fitting failed :/')

checks = [
    ("FeH", "FeH_err"),
    ("Teff",  "Teff_err"),
    ("numax", "numax_err")
]

table1 = flag_3sigma(table1,checks)

n = table1["FITFLAG"].sum()
print(f"{n}/{len(table1)} stars flagged ({100*n/len(table1):.1f}%)")

table1 = table1[['starid', 'numax_input', 'numax_err', 'G_GAIA', 'Mj_2MASS', 'Mh_2MASS', 'Mk_2MASS', 'Teff_input', 'Teff_err', 'FeH_input', 'FeH_err', 'phase', 'distance', 'distance_errp', 'distance_errm', 'radPhot',
       'radPhot_errp', 'radPhot_errm', 'massfin', 'massfin_errp',
       'massfin_errm', 'age', 'age_errp', 'age_errm','FITFLAG']]

table1['phase'] = table1['phase'].str.upper()
table1['phase'] = table1['phase'].replace({'CLUMP':'RC'})

table1.to_csv('../data/paper_tables//table_1.csv',index=False)
table1.iloc[0:10].to_csv('../data/paper_tables/table_1_first_10.csv',index=False)
```
