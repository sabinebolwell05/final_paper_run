```python
import shutil
import os
import glob
```

```python
src_root = os.path.expanduser('~/BASTA/BASTA/final_runs/output')
dst_root = os.path.expanduser('~/OneDrive - UNSW/final_runs/output')

for num in [1,2,3,4]:
    for let in 'abcd':
        name = f'table{num}_{let}'
        dst_dir = os.path.join(dst_root, name)
        os.makedirs(dst_dir, exist_ok=True)

        files = glob.glob(os.path.join(src_root, name, 'results*.ascii'))
        if not files:
            print(f'No matches in {name}')
            continue

        for f in files:
            shutil.copy(f, dst_dir)
            print(f'Copied {f} -> {dst_dir}')
```
