```python
import pandas as pd
```

```python
true_sources = pd.read_csv('../data/one_solution.csv',dtype={'source_id':str})
```

```python
#file crossmatching GaiaDR3 with kepler
keptgt_crossmatch = pd.read_csv('/Users/z5214708/Desktop/catalogs/GodoyRivera25_TableA1.csv')
keptgt_crossmatch['GaiaDR3'] = keptgt_crossmatch['GaiaDR3'].astype(str)
```

```python
#mh 2019 sample_faint
mh_2019 = pd.read_csv('/Users/z5214708/Desktop/catalogs/mh_2019.dat', skiprows=5, sep='|')
#yu 2018 sample_faint
yu_2018 = pd.read_csv('/Users/z5214708/Desktop/catalogs/yu_rgs.txt', sep='|')
mh_2019 = mh_2019[['KIC', 'nu_max']].rename(columns={'KIC': 'KICID_kep', 'nu_max': 'numax_kep'})
yu_2018 = yu_2018[['KICID', 'numax']].rename(columns={'KICID': 'KICID_kep', 'numax': 'numax_kep'})
#limit to mh not found in yu
mh_2019 = mh_2019[~mh_2019['KICID_kep'].isin(yu_2018['KICID_kep'])]
#combine into full list of oscillating kepler targets (len = 21915) 
kep_osc = pd.concat([yu_2018, mh_2019])
```

```python
#get KIC ids for stars in the crossmatch file
true_sources = true_sources.merge(keptgt_crossmatch[['KIC','GaiaDR3']], left_on='source_id',right_on='GaiaDR3', how='left')
```

```python
#remove stars that are already in yu hon
true_sources = true_sources[~true_sources['KIC'].isin(kep_osc['KICID_kep'])]
```

```python
true_sources.to_csv('../data/true_one_solution.csv', index=False)
```

```python
#now deal with mult sol
mult_sol = pd.read_csv('../data/mult_solution.csv',dtype={'source_id':str})
```

```python
#get KIC ids for stars in the crossmatch file
mult_sol = mult_sol.merge(keptgt_crossmatch[['KIC','GaiaDR3']], left_on='source_id',right_on='GaiaDR3', how='left')
#stars that are already in yu hon
in_yuhon = mult_sol[mult_sol['KIC'].isin(kep_osc['KICID_kep'])]
#remove all cases where one of the stars is in yu hon
mult_sol = mult_sol[~mult_sol['fname'].isin(in_yuhon['fname'])]
```

```python
mult_sol.to_csv('../data/true_mult_solution.csv', index=False)
```
